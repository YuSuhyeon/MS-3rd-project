```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

base_url = "https://korean.visitseoul.net/attractions?curPage={}"
headers = {
    "User-Agent": "Mozilla/5.0"
}

results = []

for page in range(1, 58):  # 1~57 페이지
    url = base_url.format(page)
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.content, "html.parser")

    elements = soup.select("div.infor-element-inner")

    for elem in elements:
        name = elem.select_one("span.title")
        name = name.get_text(strip=True) if name else ""

        desc = elem.select_one("span.small-text.text-dot-d")
        desc = desc.get_text(strip=True) if desc else ""

        rating_img = elem.select_one("span.trip-ico img")
        rating = ""
        if rating_img and rating_img.has_attr("alt"):
            # 예: '트립어드바이저 : 평점 4.4'
            rating = rating_img["alt"].split("평점")[-1].strip()

        reviews = elem.select_one("span.trip-text")
        reviews = reviews.get_text(strip=True).replace(" reviews", "") if reviews else ""

        results.append({
            "명소명": name,
            "설명": desc,
            "평점": rating,
            "리뷰수": reviews
        })

    print(f"[+] Page {page} 완료, 누적 {len(results)}개")
    time.sleep(1)  # 서버 부하 방지

# DataFrame으로 저장
df = pd.DataFrame(results)
df.to_csv("명소크롤링.csv", index=False, encoding="utf-8-sig")
print("✅ 전체 저장 완료")
```

#### 음식 제외한 파일 concat
#### 리뷰 수 내림차순 정렬 -> 중복 행 제거 -> 리뷰 없는거 제거

```python
import pandas as pd

# 사용할 파일 목록
files = [
    "비짓서울 크롤링/명소크롤링.csv",
    "비짓서울 크롤링/엔터테인먼트크롤링.csv",
    "비짓서울 크롤링/자연&관광크롤링.csv",
    "비짓서울 크롤링/쇼핑크롤링.csv"
]

# 인코딩 시도
encodings = ['utf-8-sig', 'cp949']
dfs = []
row_counts = {}

for enc in encodings:
    try:
        for file in files:
            df = pd.read_csv(file, encoding=enc)
            df['출처'] = file  # ✅ 파일명으로 출처 표시
            dfs.append(df)
        break  # 성공하면 종료
    except UnicodeDecodeError:
        dfs = []  # 실패 시 초기화 후 다음 인코딩 시도
        continue
else:
    raise ValueError("📛 인코딩 오류 발생: utf-8-sig, cp949 모두 실패")

# 파일 통합
df_all = pd.concat(dfs, ignore_index=True)

# 리뷰수 컬럼 정제 (숫자만 추출 후 float 변환)
df_all['리뷰수'] = (
    df_all['리뷰수']
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.extract(r"(\d+)", expand=False)
    .astype(float)
)

# ✅ 리뷰수 없는 항목 제거 (NaN 또는 0 이하)
before_filter = df_all.shape[0]
df_all = df_all[df_all['리뷰수'] > 0].reset_index(drop=True)
after_filter = df_all.shape[0]
print(f"🧹 리뷰수 없는 항목 제거: {before_filter - after_filter}개 제거, 남은 {after_filter}개")

# 리뷰수 기준 정렬
df_sorted = df_all.sort_values(by='리뷰수', ascending=False).reset_index(drop=True)

# 명소명이 중복되는 경우, 리뷰 수가 가장 많은 것만 남김
df_unique = df_sorted.drop_duplicates(subset='명소명', keep='first').reset_index(drop=True)

# '물품보관소'라는 단어가 명소명에 들어간 행 제거
df_cleaned = df_unique[~df_unique['명소명'].str.contains('물품보관소', na=False)].reset_index(drop=True)

# 결과 출력
print(f"🎯 리뷰수 필터링 후 전체: {df_sorted.shape[0]}행")
print(f"✅ 중복 제거 후: {df_unique.shape[0]}행")
print(f"🧹 '물품보관소' 포함 명소 제거 후: {df_cleaned.shape[0]}행")
print(df_cleaned.head(10)[['명소명', '설명', '평점', '리뷰수', '출처']])

# 저장
df_cleaned.to_csv("서울POI풀.csv", index=False, encoding="utf-8-sig")
print("📁 저장 완료: 서울POI풀.csv")
```

# 키워드 매핑

konlpy 사용해 명사 추출 -> td-idf로 상위 키워드 추출

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from konlpy.tag import Okt
import numpy as np

# === 1. 파일 불러오기 ===
df = pd.read_csv("서울POI풀.csv", encoding="utf-8-sig")
df['명소명'] = df['명소명'].fillna('')
df['설명'] = df['설명'].fillna('')

# === 2. 명사 추출 함수 정의 ===
okt = Okt()

def extract_nouns(text):
    try:
        return ' '.join(okt.nouns(text))
    except:
        return ''

# === 3. 명사 텍스트 생성 ===
df['명사_텍스트'] = (df['명소명'] + ' ' + df['설명']).apply(extract_nouns)

# === 4. TF-IDF 벡터화 ===
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df['명사_텍스트'])

# === 5. 상위 TF-IDF 키워드 추출 함수 ===
def get_top_keywords(row_index, top_n=10):
    row = X[row_index].toarray().flatten()
    top_indices = row.argsort()[::-1][:top_n]
    keywords = [vectorizer.get_feature_names_out()[i] for i in top_indices if row[i] > 0]
    return keywords

# === 6. 각 명소에 대해 상위 키워드 추출 ===
df['top_keywords'] = [get_top_keywords(i, top_n=10) for i in range(len(df))]

# === 7. 불필요한 컬럼 제거 ===
df = df.drop(columns=['명사_텍스트'])

# === 8. 저장 ===
df.to_csv("서울POI풀_키워드추출.csv", index=False, encoding="utf-8-sig")

print("✅ 키워드 컬럼 추가 완료 → 서울POI풀_키워드추출.csv 저장됨")
print(df[['명소명', 'top_keywords']].head(10))
```

```python
import pandas as pd
from collections import defaultdict
import ast

# 1. 매핑표 불러오기
mapping_df = pd.read_csv("분류_페르소나_매핑표.csv", encoding="utf-8-sig")

# 2. 매핑 리스트 (대표키워드 -> 페르소나)
keyword_to_persona = []
for _, row in mapping_df.iterrows():
    keywords = str(row['대표키워드']).split(",")
    for kw in keywords:
        keyword_to_persona.append((kw.strip(), row['연관페르소나']))

# 3. 관광지 POI 데이터 불러오기
poi_df = pd.read_csv("서울POI풀_키워드추출.csv", encoding="utf-8-sig")

# top_keywords 문자열 -> 리스트 변환
poi_df['parsed_keywords'] = poi_df['top_keywords'].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) else []
)

# 4. 매핑 함수 (부분 포함 match)
def map_persona_from_keywords(keyword_list):
    scores = defaultdict(int)
    for poi_kw in keyword_list:                 # e.g. "산책길"
        for mapping_kw, persona in keyword_to_persona:
            if mapping_kw in poi_kw:            # 부분 매칭
                scores[persona] += 1
    if scores:
        sorted_scores = sorted(scores.items(), key=lambda x: -x[1])
        return [p for p, _ in sorted_scores]
    return []

# 5. 적용
poi_df['persona_list'] = poi_df['parsed_keywords'].apply(map_persona_from_keywords)
poi_df['top1_persona'] = poi_df['persona_list'].apply(lambda x: x[0] if len(x) > 0 else None)
poi_df['top2_persona'] = poi_df['persona_list'].apply(lambda x: x[1] if len(x) > 1 else None)
poi_df['top3_persona'] = poi_df['persona_list'].apply(lambda x: x[2] if len(x) > 2 else None)

poi_df

# 6. 저장
poi_df.to_csv("서울POI_페르소나매핑.csv", index=False, encoding="utf-8-sig")
print("✅ 매핑 완료: 서울POI_페르소나매핑.csv")
```
